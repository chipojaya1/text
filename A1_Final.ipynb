#### Question 1: Validate an Email Address

```python
import re
def is_valid_email(email: str) -> bool:
    """
    Checks if the input string is a valid email address using a regular expression.
    A valid email should follow the format: username@domain.extension.

    It allows letters, numbers, and common symbols like '.', '-', and '_'
    in the username and domain, followed by a 2-4 letter extension.
    """
    # Pattern breakdown:
    # ^                     # Start of the string (Anchor)
    # [\w.-]+               # Username: one or more (+) word characters (\w, including [a-zA-Z0-9_]), dot (.), or dash (-)
    # @                     # Literal '@' symbol
    # [\w-]+                # Domain name: one or more word characters or dash (-)
    # \.                    # Literal dot (.): must be escaped
    # [a-zA-Z]{2,4}         # Extension: 2 to 4 letters (e.g., com, org, co, info)
    # $                     # End of the string (Anchor)

    pattern = r'^[\w.-]+@[\w-]+\.[a-zA-Z]{2,4}$'

    return bool(re.match(pattern, email))

# Example Usage (Testing the function)
print(f"chipo.jaya@email.gwu.edu: {is_valid_email('chipo.jaya@email.gwu.edu')}")
print(f"chipo.jaya@gwmail.gwu.edu: {is_valid_email('chipo.jaya@gwmail.gwu.edu')}")
print(f"chipo.jaya@gwu.edu: {is_valid_email('chipo.jaya@gwu.edu')}")
```

#### Question 2: Extract All Phone Numbers from a Text

```python
import re

# Question 2: Extract All Phone Numbers from a Text
def extract_phone_numbers(text):
    """
    Finds and returns all phone numbers in the format XXX-XXX-XXXX
    from the input text using a regular expression.
    """
    # Pattern breakdown:
    # \d{3}       # Matches exactly 3 digits
    # -           # Matches the literal dash
    # \d{3}       # Matches exactly 3 digits
    # -           # Matches the literal dash
    # \d{4}       # Matches exactly 4 digits

    pattern = r'\d{3}-\d{3}-\d{4}'

    # re.findall returns a list of all non-overlapping matches of the pattern.
    return re.findall(pattern, text)

# Example Usage (Testing the function)
sample_text = "Contact us at 123-456-7890 or 987-654-3210."

found_numbers = extract_phone_numbers(sample_text)

print(f"Input Text: \"{sample_text}\"")
print(f"Extracted Phone Numbers: {found_numbers}")
```

#### Question 3: Replace All Digits with a Hash Symbol

```python
import re

# Question 3: Replace ALL Digits with a Hash Symbol
def replace_digits(text: str) -> str:
    """
    Replaces every occurrence of a digit (0-9) in the input string
    with the '#' symbol using a regular expression.
    """
    # Pattern breakdown:
    # # \d Matches any single digit character (equivalent to [0-9])
    pattern = r'\d'

    # re.sub(pattern, replacement, text) replaces all matches of the pattern with the replacement string.
    return re.sub(pattern, '#', text)

# Example Usage (Testing the function)
sample_text = "I have 30 apples and 5000 oranges."
result = replace_digits(sample_text)

print(f"Original Text: {sample_text}")
print(f"Resulting Text: {result}")
```

#### Question 4: Find All Words Starting with a Capital Letter

```python
import re

# Question 4: Find All Words Starting with a Capital Letter
def find_capital_words(text):
    """
    Finds and returns a list of all words in the input text that start
    with an uppercase letter, using a regular expression.
    """
    # Pattern breakdown:
    # \b          # Word boundary (Start of a word)
    # [A-Z]       # Matches exactly one uppercase letter
    # \w* # Matches zero or more word characters (letters, numbers, or underscore)
    # \b          # Word boundary (End of a word)

    pattern = r'\b[A-Z]\w*\b'

    return re.findall(pattern, text)

# Example Usage (Testing the function)
sample_text = "The quick Brown fox Jumps over the Lazy dog. I am A coder."

found_words = find_capital_words(sample_text)

print(f"Original Text: {sample_text}")
print(f"Extracted Capital Words: {found_words}")

# Example Output for demonstration
# Original Text: The quick Brown fox Jumps over the Lazy dog. I am A coder.
# Extracted Capital Words: ['The', 'Brown', 'Jumps', 'Lazy', 'I', 'A']
```

#### Question 5: Check if a String is a Valid Date (DD-MM-YYYY)

```python
import re

# Question 5: Check if a String is a Valid Date (DD-MM-YYYY)
def is_valid_date(date_str: str) -> bool:
    """
    Checks if the input string is a valid date in the format DD-MM-YYYY,
    enforcing that the day is 01-31 and the month is 01-12.
    """
    # Pattern breakdown:
    # ^                        # Start anchor
    # (0[1-9]|[12]\d|3[01])    # Day (DD): Group 1 (01-09) OR (10-29) OR (30-31)
    # -                        # Literal dash separator
    # (0[1-9]|1[0-2])          # Month (MM): Group 2 (01-09) OR (10-12)
    # -                        # Literal dash separator
    # \d{4}                    # Year (YYYY): Matches exactly 4 digits
    # $                        # End anchor

    pattern = r'^(0[1-9]|[12]\d|3[01])-(0[1-9]|1[0-2])-\d{4}$'

    # re.match checks for a match only at the beginning of the string.
    # We use bool() to return True/False based on whether a match object is found.
    return bool(re.match(pattern, date_str))

# Example Usage (Testing the function)
valid_date = "20-02-1994"
invalid_day = "32-02-1994"
invalid_month = "10-13-1994"
invalid_format = "1-1-194"

print(f"'{valid_date}' is valid: {is_valid_date(valid_date)}")
print(f"'{invalid_day}' is valid: {is_valid_date(invalid_day)}")
print(f"'{invalid_month}' is valid: {is_valid_date(invalid_month)}")
print(f"'{invalid_format}' is valid: {is_valid_date(invalid_format)}")
```

#### Bonus: Improve Question 5 to reject invalid dates like '31-02-2025' or '30-02-2025.'

```python
import re

# Question 5 (February Only): Check if a String is a Valid February Date (DD-02-YYYY)
def is_valid_february_date(date_str: str) -> bool:
    """
    Checks if the input string is a valid date in the format DD-02-YYYY,
    enforcing that the day is 01-28, using a regular expression.

    This function specifically targets non-leap-year February dates.
    """

    # Pattern breakdown for DD-02-YYYY (DD limited to 01-28):
    # ^                      # Start anchor
    # (0[1-9]|1\d|2[0-8])    # Day (DD): Group 1 (01-09) OR (10-19) OR (20-28)
    # -                      # Literal dash separator
    # 02                     # Month (MM): Matches '02' literally
    # -                      # Literal dash separator
    # \d{4}                  # Year (YYYY): Matches exactly 4 digits
    # $                      # End anchor

    pattern = r'^(0[1-9]|1\d|2[0-8])-02-\d{4}$'

    # re.match checks for a match only at the beginning of the string.
    # The start (^) and end ($) anchors ensure the whole string must match.
    return bool(re.match(pattern, date_str))

# Example Usage (Testing the function)
valid_feb = "28-02-1994"       # Valid February (non-leap year)
valid_early_feb = "05-02-1994" # Valid early February
invalid_29_in_feb = "29-02-1994" # Invalid: 29 days (rejects all leap years)
invalid_30_in_feb = "30-02-1994" # Invalid: 30 days in February
invalid_31_in_feb = "31-02-1994" # Invalid: 31 days in February

print(f"'{valid_feb}' is valid: {is_valid_february_date(valid_feb)}")
print(f"'{valid_early_feb}' is valid: {is_valid_february_date(valid_early_feb)}")
print(f"'{invalid_29_in_feb}' is valid: {is_valid_february_date(invalid_29_in_feb)}")
print(f"'{invalid_30_in_feb}' is valid: {is_valid_february_date(invalid_30_in_feb)}")
print(f"'{invalid_31_in_feb}' is valid: {is_valid_february_date(invalid_31_in_feb)}")
```
